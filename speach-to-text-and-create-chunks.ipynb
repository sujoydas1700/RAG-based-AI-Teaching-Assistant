```python
import whisper
# download model or load
model = whisper.load_model('large-v2')
```

```python
import os
os.chdir("/content/drive/MyDrive/RAG based AI model Project")
print("current dir :-", os.getcwd())
```

```python
result = model.transcribe(audio="8_sample.mp3", language="hi", task="translate", word_timestamps=False)
```

```python
print(result)
```

```python
print(result["segments"])
```

```python
chunks = []
for segment in result["segments"]:
  chunks.append({"start":segment["start"],"end":segment["end"],"text":segment["text"]})

print(chunks)
```

```python
# testing on small sample data
import json
with open("sample_output.json","w") as f:
  json.dump(chunks,f)
```

```python
audios = os.listdir("audios")
# audios = ["8_sample.mp3"]
for audio in audios:
  num = audio[0:1]
  name = audio[2:-4]
  print(num, name)
  result = model.transcribe(audio=f"audios/{audio}", language="hi", task="translate", word_timestamps=False)

  chunks = []
  for segment in result["segments"]:
    chunks.append({"number":num,"title":name,"start":segment["start"],"end":segment["end"],"text":segment["text"]})
  chunks_with_metadata = {"chunks":chunks,"text":result["text"]}
  with open(f"jsons/{audio}.json","w") as f:
    json.dump(chunks_with_metadata,f)
  print(f"{audio} done")
```
